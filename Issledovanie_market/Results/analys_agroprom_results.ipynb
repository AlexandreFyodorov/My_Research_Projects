# Аналитический отчёт по результатам фундаментального скрининга сектора Агропром и химии

**Дата анализа:** 16 марта 2026 г.  
**Выборка:** 15 компаний агропромышленного комплекса и химической промышленности, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **PHOR, GCHE, NKHP, AKRN, KAZT, ABRD, KLVZ, RAGR, AQUA**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского агропромышленного комплекса и химической отрасли, представленных в файле и отобранных пользователем. В выборку вошли ведущие производители минеральных удобрений, крупные агрохолдинги, производители продуктов питания и напитков, а также компании, связанные с переработкой сельхозпродукции и аквакультурой.

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

## 3. Детальный анализ фаворитов

### 3.1. PHOR (ФосАгро)

**Динамика ключевых мультипликаторов (2020-2025):**

| Показатель | 2020 | 2021 | 2022 | 2023 | 2024 | 2025 | LTM |
|------------|------|------|------|------|------|------|-----|
| P/E | 9.37 | 5.81 | 4.54 | 8.20 | 8.20 | н/д | 8.04 |
| P/B | 3.83 | 4.64 | 4.24 | 5.74 | 5.10 | н/д | 4.34 |
| EV/EBITDA | 6.50 | 4.71 | 3.90 | 5.86 | 6.47 | н/д | 5.42 |
| NetDebt/EBITDA | 1.76 | 0.77 | 0.69 | 1.20 | 1.81 | н/д | 1.14 |

**Анализ динамики:**  
В 2021-2022 годах наблюдалось существенное улучшение мультипликаторов благодаря взрывному росту цен на удобрения на мировых рынках. P/E снизился до исторических минимумов (4.5-5.8), что указывало на глубокую недооценку при высокой прибыли. В 2023 году произошла коррекция цен на удобрения, что отразилось в росте P/E до 8.2. В 2024-2025 годах показатели стабилизировались: P/E держится в диапазоне 8-9, P/B вырос до 5.1 (дорогая оценка относительно баланса), NetDebt/EBITDA увеличился до 1.8, но остаётся в комфортной зоне.

**Свежие данные и перспективы:**  
ФосАгро демонстрирует высокое кредитное качество (рейтинг ruAAA). По данным на февраль 2026 года, дивидендные выплаты составили 273 руб. на акцию (доходность около 4% по последним выплатам). Текущие мультипликаторы по данным Интерфакса: P/E 8.2, EV/EBITDA 6.26, Debt/EBITDA 1.12. Стратегия государства по снижению доли зерна в агроэкспорте до 25% к 2030 году в пользу продукции с высокой добавленной стоимостью создаёт долгосрочный спрос на удобрения.

**Вывод по тикеру:**  
ФосАгро – высококачественный эмитент с сильным кредитным рейтингом и стабильными дивидендами. Текущие мультипликаторы (P/E ~8) указывают на справедливую оценку. Рекомендуется для консервативных инвесторов, ориентированных на долгосрочный рост и дивиденды.

## 5. Выводы и рекомендации

### 5.1. Лидеры по дивидендной доходности

- **NKHP (НКХП)** – лидер по дивидендной доходности (4.4%) среди анализируемых компаний. Уникальная компания с отрицательным долгом и потенциалом роста на расширении мощностей. Рекомендуется для инвесторов, делающих ставку на восстановление экспорта зерна.



- **PHOR (ФосАгро)** – стабильные дивиденды (4.16%), высокая рентабельность (ROE 53%), умеренный долг. Наилучший выбор для консервативных инвесторов.



- **AKRN (Акрон)** – дивидендная доходность 4.35% при низком долге, но высокая волатильность мультипликаторов. Рекомендуется для инвесторов, готовых к циклическим колебаниям.



### 5.2. Наиболее привлекательные по фундаментальным показателям

- **AQUA (Русская Аквакультура)** – ставка на восстановительный цикл после биологического кризиса. Рекордное зарыбление 2025 года создаёт потенциал роста продаж в 2026-2027.



- **KAZT (КуйбышевАзот)** – недооценка по P/B (0.82) при низком долге и стабильном бизнесе.



### 5.3. Умеренный интерес

- **GCHE (Черкизово)** – стабильный бизнес, но отсутствие дивидендов в 2025 году и долговая нагрузка требуют осторожности.



### 5.4. Высокорисковые / требующие осторожности

- **RAGR (Русагро)** – высокий корпоративный риск из-за смены контроля, отсутствие дивидендов.
- **ABRD (Абрау-Дюрсо)** – высокая долговая нагрузка (NetDebt/EBITDA 3.01), отсутствие дивидендов.
- **KLVZ (КЛВЗ Кристалл)** – дорогая оценка (P/E 27.9) при низкой дивидендной доходности.

## 6. Перспективы сектора на 2026-2029 гг.

- **Минеральные удобрения (PHOR, AKRN)** – останутся бенефициарами глобального спроса на продовольствие. Стратегия импортозамещения создаёт долгосрочный спрос.
- **Перевалка зерна (NKHP)** – восстановление экспорта при ослаблении рубля и снятии квот. Расширение мощностей создаёт потенциал роста.
- **Аквакультура (AQUA)** – восстановление после биологического кризиса, потенциал роста при нормализации биологических циклов и возможный возврат к двузначной дивидендной доходности.
- **Риски:** высокая ключевая ставка, волатильность рубля, экспортные пошлины и квоты, биологические риски, корпоративные конфликты.

```python
# =============================================================================
# БЛОК АНАЛИЗА ДИНАМИКИ МУЛЬТИПЛИКАТОРОВ ПО СЕКТОРУ (2020-2025)
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import os
import warnings
warnings.filterwarnings('ignore')

# -------------------- НАСТРОЙКИ --------------------
YEARS = [2020, 2021, 2022, 2023, 2024, 2025]  # анализируемые годы
INPUT_YEARLY_FILE = 'C:/Users/1/Documents/TRADE_2026/ScriptsAnalizTorgy/df_fund_yearly.xlsx'
SECTOR_FILE = 'C:/Users/1/Documents/TRADE_2026/ScriptsAnalizTorgy/Sector_updated.xlsx'  # файл с секторами
OUTPUT_PATH = 'C:/Users/1/Documents/TRADE_2026/ScriptsAnalizTorgy/Analitika/'
SECTOR_LIST = [
    'Агропром, химия'
]

# Список показателей для анализа (расширенный, включая дивидендную доходность)
METRICS = [
    'DivYield_ao', 'DivYield_ap', 'Revenue', 'OperatingProfit', 'EBITDA', 'NetIncome',
    'CAPEX', 'FCF', 'OperExpenses', 'Interest',
    'Assets', 'Debt', 'Cash', 'NetDebt',
    'Price', 'Shares', 'MarketCap', 'EPS', 'BVPS',
    'ROE_raw', 'ROA_raw', 'PE_raw', 'PS_raw', 'PB_raw',
    'EVEBITDA_raw', 'DebtEBITDA_raw'
]

# -------------------- ЗАГРУЗКА ДАННЫХ --------------------
print("Загрузка годовых данных...")
yearly_dfs = {}
for year in YEARS:
    try:
        df_year = pd.read_excel(INPUT_YEARLY_FILE, sheet_name=str(year), index_col=0, engine='openpyxl')
        # Убираем возможный дублирующийся индекс, если колонка 'A' попала как данные
        if 'A' in df_year.columns:
            df_year = df_year.drop(columns=['A'])

        # --- Добавление дивидендной доходности, если есть Dividend и Price ---
        if 'Dividend' in df_year.columns and 'Price' in df_year.columns:
            mask = (df_year['Price'] > 0) & df_year['Dividend'].notna()
            df_year.loc[mask, 'DivYield'] = df_year.loc[mask, 'Dividend'] / df_year.loc[mask, 'Price'] * 100
            # print(f"    Рассчитана DivYield для {year} года")
        else:
            # Если нет данных, колонка всё равно может отсутствовать – это нормально
            pass

        yearly_dfs[year] = df_year
        print(f"  {year}: {df_year.shape[0]} строк, {df_year.shape[1]} колонок")
    except Exception as e:
        print(f"  Ошибка загрузки листа {year}: {e}")
        yearly_dfs[year] = pd.DataFrame()

# -------------------- ЗАГРУЗКА СЕКТОРОВ --------------------
print("\nЗагрузка секторов...")
try:
    sector_df = pd.read_excel(SECTOR_FILE, sheet_name='Sector', engine='openpyxl')
    sector_df = sector_df[['Код инструмента', 'Категории']].copy()
    sector_df.dropna(inplace=True)
    sector_df['Код инструмента'] = sector_df['Код инструмента'].astype(str).str.strip()
    sector_df = sector_df.drop_duplicates(subset=['Код инструмента'])
    sector_mapping = dict(zip(sector_df['Код инструмента'], sector_df['Категории']))
    print(f"Загружено {len(sector_mapping)} записей о секторах.")
except Exception as e:
    print(f"Ошибка загрузки секторов: {e}")
    sector_mapping = {}

def get_sector(ticker):
    return sector_mapping.get(str(ticker).strip(), 'Другое')

# -------------------- ОТБОР ТИКЕРОВ ПО СЕКТОРУ --------------------
all_tickers = set()
for df in yearly_dfs.values():
    all_tickers.update(df.index)
all_tickers = sorted(all_tickers)

# Сопоставляем тикеры с секторами
ticker_sector = {ticker: get_sector(ticker) for ticker in all_tickers}

# Выбираем тикеры, входящие в SECTOR_LIST
selected_tickers = [t for t in all_tickers if ticker_sector[t] in SECTOR_LIST]
print(f"\nТикеры сектора {SECTOR_LIST}: {selected_tickers}")

if not selected_tickers:
    raise ValueError("Нет тикеров для выбранного сектора. Проверьте SECTOR_LIST и файл с секторами.")

# -------------------- ПОСТРОЕНИЕ ГРАФИКОВ ДЛЯ КАЖДОГО ПОКАЗАТЕЛЯ --------------------
os.makedirs(OUTPUT_PATH, exist_ok=True)
timestamp = datetime.now().strftime('%d%m%y_%H%M')

# Для каждого показателя создадим отдельный рисунок
for metric in METRICS:
    print(f"\nОбработка показателя: {metric}")

    # Собираем данные по годам для выбранных тикеров
    data = []
    for year in YEARS:
        df_year = yearly_dfs.get(year)
        if df_year is None or df_year.empty:
            continue
        if metric not in df_year.columns:
            # Если метрики нет, пропускаем (например, DivYield может быть не во всех годах)
            continue
        # Извлекаем значения для отобранных тикеров
        for ticker in selected_tickers:
            if ticker in df_year.index:
                val = df_year.loc[ticker, metric]
                if pd.notna(val) and val not in [np.inf, -np.inf]:
                    data.append({
                        'Год': year,
                        'Тикер': ticker,
                        metric: val
                    })
    if not data:
        print(f"  Нет данных для {metric}, пропускаем")
        continue

    df_plot = pd.DataFrame(data)

    # Определяем количество тикеров и подбираем размер графика
    n_tickers = len(df_plot['Тикер'].unique())
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # --- Левый график: линейный (динамика по годам для каждого тикера) ---
    ax1 = axes[0]
    sns.lineplot(data=df_plot, x='Год', y=metric, hue='Тикер', marker='o', ax=ax1, palette='tab10')
    ax1.set_title(f'Динамика {metric} по годам\nСектор: {SECTOR_LIST[0]}', fontsize=12)
    ax1.set_xlabel('Год')
    ax1.set_ylabel(metric)
    ax1.legend(title='Тикер', bbox_to_anchor=(1.05, 1), loc='upper left')
    ax1.grid(True, alpha=0.3)

    # --- Правый график: столбчатая диаграмма (сравнение тикеров по годам) ---
    ax2 = axes[1]
    # Сгруппируем по годам и тикерам, построим grouped bar chart
    pivot_df = df_plot.pivot(index='Год', columns='Тикер', values=metric)
    pivot_df.plot(kind='bar', ax=ax2, width=0.8)
    ax2.set_title(f'Сравнение {metric} по годам\nСектор: {SECTOR_LIST[0]}', fontsize=12)
    ax2.set_xlabel('Год')
    ax2.set_ylabel(metric)
    ax2.legend(title='Тикер', bbox_to_anchor=(1.05, 1), loc='upper left')
    ax2.grid(axis='y', alpha=0.3)

    plt.suptitle(f'Анализ показателя: {metric}', fontsize=14, fontweight='bold')
    plt.tight_layout()

    # Сохраняем график
    filename = f'{metric}_dynamics_{SECTOR_LIST[0].replace(" ", "_")}_{timestamp}.png'
    filepath = os.path.join(OUTPUT_PATH, filename)
    plt.savefig(filepath, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  График сохранён: {filepath}")

    # Вывод описания показателя
    descriptions = {
        'Revenue': 'Выручка — общий доход компании от реализации продукции/услуг.',
        'OperatingProfit': 'Операционная прибыль — прибыль от основной деятельности до вычета процентов и налогов.',
        'EBITDA': 'Прибыль до вычета процентов, налогов, износа и амортизации — показатель операционной эффективности.',
        'NetIncome': 'Чистая прибыль — конечный финансовый результат.',
        'CAPEX': 'Капитальные затраты — инвестиции в основные средства.',
        'FCF': 'Свободный денежный поток — денежные средства, доступные после инвестиций.',
        'OperExpenses': 'Операционные расходы — затраты на основную деятельность.',
        'Interest': 'Процентные расходы — затраты на обслуживание долга.',
        'Assets': 'Активы — всё, чем владеет компания.',
        'Debt': 'Долг — общая сумма заёмных средств.',
        'Cash': 'Денежные средства и их эквиваленты.',
        'NetDebt': 'Чистый долг (Долг минус Денежные средства).',
        'Price': 'Цена акции.',
        'Shares': 'Количество акций в обращении.',
        'MarketCap': 'Рыночная капитализация (Цена × Количество акций).',
        'EPS': 'Прибыль на акцию.',
        'BVPS': 'Балансовая стоимость на акцию.',
        'ROE_raw': 'Рентабельность собственного капитала (Чистая прибыль / Собственный капитал).',
        'ROA_raw': 'Рентабельность активов (Чистая прибыль / Активы).',
        'PE_raw': 'Цена / Прибыль — мультипликатор, показывающий сколько лет потребуется для окупаемости инвестиции за счёт прибыли.',
        'PS_raw': 'Цена / Выручка — оценка стоимости компании относительно выручки.',
        'PB_raw': 'Цена / Балансовая стоимость — сравнение рыночной и балансовой стоимости.',
        'EVEBITDA_raw': 'Стоимость компании (EV) / EBITDA — показатель сравнительной оценки.',
        'DebtEBITDA_raw': 'Чистый долг / EBITDA — показатель долговой нагрузки.',
        'DivYield_ao': 'Дивидендная доходность (обычные акции) = Дивиденд на акцию / Цена акции × 100%. Показывает, сколько процентов от текущей цены акции составляют годовые дивиденды.',
        'DivYield_ap': 'Дивидендная доходность (привелегированные акции) = Дивиденд на акцию / Цена акции × 100%. Показывает, сколько процентов от текущей цены акции составляют годовые дивиденды.'
    }
    desc = descriptions.get(metric, 'Нет описания.')
    print(f"  Описание: {desc}\n")

print(f"\nАнализ завершён. Все графики сохранены в папку: {OUTPUT_PATH}")
```
