# Аналитический отчёт по результатам фундаментального скрининга IT-компаний

**Дата анализа:** 13 марта 2026 г.  
**Выборка:** 18 компаний IT-сектора, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):**  
- Группа 1 (наилучшая дивидендная доходность): **MTSS, DIAS, TTLK**.  
- Группа 2 (хорошие ROE/ROA, но с долгами, в порядке роста NetDebt): **T, DATA, SOFT, AFKS, RTKM, YDEX, ELMT, ASTR, CNTLP**.



## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского IT-сектора, отобранных пользователем, с акцентом на:
- компании с максимальной дивидендной доходностью (включая MTSS, где высокая дивидендная доходность сочетается со значительным долгом, но сохраняется дивидендный ориентир);
- компании, демонстрирующие высокую рентабельность капитала (ROE) и активов (ROA), имеющие положительный чистый долг (NetDebt), ранжированные по возрастанию долга.

В работе использованы данные из предоставленного файла, а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

**Анализ ситуации и перспективы:**  
МТС остаётся классической дивидендной историей на российском рынке. Согласно данным Т-Банка, компания демонстрирует устойчивый рост выручки (двузначные темпы в 2025 году) благодаря индексации тарифов и развитию экосистемы (реклама, финтех, IT-сервисы) [citation:1]. Менеджмент подтверждает намерение выплачивать дивиденды в размере 35 руб./акцию в ближайшие годы, что обеспечивает доходность 15-16% [citation:1][citation:8].

**Ключевые факторы на 2026-2029 гг.:**
- **Долговая нагрузка:** около 2/3 долга имеют плавающую ставку. Снижение ключевой ставки ЦБ (ожидаемое во 2П 2026) уменьшит расходы на обслуживание и увеличит свободный денежный поток [citation:1].
- **Риски:** высокий долг (741 млрд руб.) и отрицательный собственный капитал создают напряжённость. Критики отмечают, что дивиденды могут выплачиваться за счёт новых заимствований, а свободный денежный поток без учёта банка составляет всего 0.4 млрд руб. [citation:8].
- **Драйверы роста:** возможные IPO дочерних компаний (Adtech, Erion, Юрент) и пересмотр дивидендной политики при стабилизации долга [citation:1].

**Вердикт:** привлекательна для дивидендных инвесторов, готовых принимать долговые риски. Рекомендуется держать до дивидендной отсечки с контролем долговой ситуации.

## 4. Выводы и рекомендации

### 4.1. Фавориты первой группы
- **MTSS** – остаётся дивидендной легендой, но долговая нагрузка требует постоянного мониторинга. Рекомендуется для дивидендных портфелей с горизонтом до 1 года и готовностью к волатильности.
- **DIAS** – идеальное сочетание отсутствия долга, высокой рентабельности и дивидендной доходности. Рекомендуется для долгосрочных консервативных вложений.
- **TTLK** – недооценённая компания с отрицательным долгом и стабильными дивидендами. Рекомендуется для стратегии «купи и держи».

### 4.2. Вторая группа (с долговой нагрузкой)
Наиболее перспективные среди них:
- **T** – уникальная компания с отрицательным долгом, высокой рентабельностью и кредитным рейтингом ruA+. Рекомендуется как якорный актив в портфеле.
- **YDEX** – лидер роста с низким долгом и высокой рентабельностью, но дорогой по P/B. Рекомендуется для инвесторов, готовых платить за качество.
- **ASTR** – сверхвысокая рентабельность при низком долге, но экстремальный P/B требует выборочного подхода. Рекомендуется для агрессивных инвесторов.
- **ELMT** – хороший представитель сектора микроэлектроники с умеренной оценкой и низким долгом. Рекомендуется для диверсификации.

Аутсайдеры: **SOFT** (невыполнение прогнозов, снижение доверия), **AFKS** (негативный прогноз АКРА, риск снижения покрытия процентов).

### 4.3. Перспективы сектора на 2026-2029 гг.
- **IT-сектор** остаётся драйвером роста благодаря импортозамещению и цифровизации экономики.
- **Телекомы** (MTSS, TTLK) выиграют от снижения ключевой ставки, но долговая нагрузка остаётся ключевым риском.
- **Финтех (T)** продолжит экспансию, поддерживаемый высокой ликвидностью и доверием рынка.
- Риски: макроэкономическая нестабильность, регуляторные изменения (ФАС, налоги), волатильность рубля.

### 4.4. Когда произвести очередной скрининг?
Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:
- динамику долговой нагрузки MTSS и AFKS;
- выполнение прогнозов по выручке SOFT и ASTR;
- изменение дивидендных политик;
- макроэкономические сигналы (ставка ЦБ, инфляция).

## 5. Общий вердикт

Пользователь сформировал качественную выборку, отражающую основные тренды IT-сектора. Наиболее сбалансированными представляются **T** и **DIAS**. Для дивидендных инвесторов интересны **MTSS** и **TTLK**. Для агрессивного роста – **YDEX** и **ASTR**. Остальные компании могут использоваться для диверсификации с учётом их специфических рисков.
