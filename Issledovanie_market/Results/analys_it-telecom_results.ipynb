# Аналитический отчёт по результатам фундаментального скрининга IT-сектора

**Дата анализа:** 17 марта 2026 г.  
**Выборка:** 10 компаний IT-сектора и смежных областей (телеком, финтех, разработка ПО), прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **MTSS, DIAS, TTLK, T, DATA, AFKS, RTKM, YDEX, ELMT, ASTR**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности ведущих компаний российского IT-сектора, представленных в файле и отобранных пользователем. В выборку вошли крупнейшие телекоммуникационные операторы, разработчики программного обеспечения, финтех-платформы, а также холдинги с активами в высокотехнологичных отраслях.

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

Компания заявляет, что планов уходить с биржи нет, но рынок оценивает риски высоко.

ELMT – высокорисковый актив с неопределённым будущим в публичном пространстве. Потенциальный делистинг и слабая динамика после IPO создают значительные риски. Рекомендуется только для спекулятивных инвесторов с пониманием корпоративных рисков.

## 5. Выводы и рекомендации

### 5.1. Лидеры по дивидендной доходности

- **MTSS (МТС)** – лидер по дивидендной доходности (15.5%) с утверждённой дивидендной политикой не менее 35 руб. на акцию. Высокий долг остаётся основным риском, но операционный бизнес стабилен.


- **T (Т-Технологии)** – дивидендная доходность 4.28% при крайне низкой оценке (P/E 5.9) и чистой денежной позиции. Потенциал роста котировок.


- **TTLK (Таттелеком)** – стабильные дивиденды (4.96%), отрицательный долг и недооценка по P/B (0.95). Защитный актив с умеренным апсайдом.


### 5.2. Наиболее привлекательные по фундаментальным показателям

- **YDEX (Яндекс)** – сочетание высоких темпов роста (выручка +32%), улучшения рентабельности (ROE 40%) и начала дивидендных выплат. Уникальные рыночные позиции.


- **ASTR (Группа Астра)** – экстремально низкий P/E (10.5) для компании с рентабельностью 67% и минимальным долгом. Главный бенефициар импортозамещения в ПО.


- **T (Т-Технологии)** – низкая оценка, высокая рентабельность, чистые деньги на балансе.


### 5.3. Умеренный интерес

- **DIAS (Диасофт)** – качественный бизнес, но замедление роста и нейтральные прогнозы аналитиков ограничивают потенциал.
- **DATA (Аренадата)** – перспективный сегмент, но высокая волатильность и нестабильные показатели.
- **RTKM (Ростелеком)** – стабильный дивидендный актив, но текущий P/E выше исторических значений.


### 5.4. Высокорисковые / требующие осторожности

- **ELMT (ГК «Элемент»)** – высокий риск делистинга, слабая динамика после IPO, корпоративная неопределённость.
- **AFKS (АФК «Система»)** – высокий долг, негативный прогноз АКРА, сложная холдинговая структура.

## 6. Перспективы сектора на 2026-2029 гг.

- **Импортозамещение ПО** остаётся ключевым драйвером для ASTR, DIAS, DATA. Государственные и корпоративные заказчики продолжат переход на отечественное ПО.
- **Телекоммуникации** (MTSS, RTKM, TTLK) – стабильный денежный поток, дивидендные выплаты, но ограниченный потенциал роста выручки.
- **Финтех** (T) – консолидация рынка, развитие экосистем, высокая рентабельность.
- **Микроэлектроника** (ELMT) – стратегическое направление, но высокая капиталоёмкость и зависимость от господдержки.
- **Ключевые драйверы:** снижение ключевой ставки (позитивно для закредитованных MTSS и AFKS), продолжение цифровизации, господдержка IT-отрасли.
- **Риски:** высокая ключевая ставка, регуляторные изменения, усиление конкуренции, корпоративные риски (ELMT, AFKS).

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- разрешение корпоративной ситуации в ELMT (риск делистинга);
- финансовые результаты YDEX и подтверждение дивидендных выплат;
- динамику долговой нагрузки MTSS и AFKS;
- темпы роста ASTR и сохранение рентабельности;
- макроэкономические сигналы (ключевая ставка ЦБ, инфляция, объёмы кредитования);
- новые IPO в IT-секторе и их влияние на оценки публичных компаний.
