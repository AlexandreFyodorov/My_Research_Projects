# Аналитический отчёт по результатам фундаментального скрининга сырьевого сектора (металлы и добыча)

**Дата анализа:** 17 марта 2026 г.  
**Выборка:** 12 компаний сырьевого сектора (чёрная и цветная металлургия, золотодобыча, уголь, удобрения), прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **LNZL, NLMK, MAGN, ALRS, SELG, CHMF, UGLD, MTLR, PLZL, GMKN, URKZ, BRZL**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского сырьевого сектора, представленных в файле и отобранных пользователем. В выборку вошли ведущие производители стали, золота, алмазов, угля, удобрений, а также предприятия по добыче и переработке минерального сырья. Сектор характеризуется высокой цикличностью, зависимостью от мировых цен на сырьё и курса рубля, что определяет волатильность финансовых показателей.

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

### 3.1. LNZL (Лензолото)

| NetDebt/EBITDA | -3.27 | - | - | - | - | - |
| ROE, % | 27.5 | 0.3 | 50.4 | 7.9 | 15.1 | 15.1 |
| ROA, % | 27.4 | 0.2 | 39.1 | 6.1 | 9.1 | 9.1 |

**Анализ динамики:**  
Компания демонстрирует экстремальную волатильность мультипликаторов из-за низкой капитализации и нестабильной чистой прибыли. В 2021 и 2023 годах P/E достигал аномальных значений (сотни единиц) на фоне падения прибыли. В LTM показатель нормализовался до 53.2, что всё ещё высоко, но сопровождается нулевым долгом (отрицательный чистый долг) и стабильной рентабельностью капитала около 15%.

**Свежие данные и перспективы:**  
Дивидендная доходность по LTM составляет 55.3% – одна из самых высоких в выборке. Компания сохраняет денежную подушку (NetDebt отрицательный). Основной риск – малая ликвидность и зависимость от цен на золото. При сохранении благоприятной конъюнктуры золота и стабилизации прибыли оценка может оставаться привлекательной.

**Вывод по тикеру:**  
LNZL – интересная дивидендная история с нулевым долгом, но требует осторожности из-за высокой волатильности прибыли и малого free float. Рекомендуется для диверсифицированных портфелей с ориентацией на дивиденды.

**Свежие данные и перспективы:**  
Дивиденды отсутствуют. Компания чувствительна к цене золота, но низкий долг и дисконт к балансу создают потенциал для роста.

**Вывод по тикеру:**  
BRZL – недооценённый актив с низким долгом. Интересен для долгосрочных инвесторов, делающих ставку на рост цен на золото.

## 5. Выводы и рекомендации

### 5.1. Лидеры по дивидендной доходности

- **PLZL (Полюс)** – стабильный дивидендный плательщик (4.4%) с комфортным долгом и высокой рентабельностью. Лучший выбор в секторе золотодобычи для консервативных инвесторов.



### 5.2. Наиболее привлекательные по фундаментальным показателям

- **NLMK (НЛМК)** – P/B 0.75, отрицательный долг, стабильный бизнес. Ожидание возобновления дивидендов.


- **BRZL (Бурятзолото)** – P/B 0.57, низкий долг, стабильная рентабельность. Ставка на рост цен на золото.


### 5.3. Умеренный интерес

- **ALRS (АЛРОСА)** – недооценка по P/B, но высокий долг и санкционные риски.


- **CHMF (Северсталь)** – сильный баланс, но высокий P/E и отсутствие дивидендов.


- **GMKN (Норникель)** – уникальные активы, но текущая оценка не дешёвая (P/E 14.3).


### 5.4. Высокорисковые / требующие осторожности

- **MTLR (Мечел)** – критический долг, отрицательный капитал, высокие риски.
- **SELG (Селигдар)** – высокий долг (NetDebt/EBITDA 4.15), несмотря на высокую рентабельность.
- **UGLD (Распадская)** – чувствительность к ценам на уголь, высокая оценка P/B.
- **MAGN (ММК)** – временная убыточность, но низкий долг создаёт потенциал для циклического разворота.
- **LNZL (Лензолото)** – уникальная компания c нулевым долгом. Однако высокий P/E (53) и малая ликвидность требуют осторожности. Активы проданы, дивиденды выплачены, в компании уже почти ничего не осталось, ни в коем случае нельзя покупать (22.09.2021). Подходит для дивидендных инвесторов с пониманием рисков.

## 6. Перспективы сектора на 2026-2029 гг.

- **Металлургия** будет зависеть от мировых цен на сталь, спроса со стороны строительства и машиностроения, а также от экспортных пошлин и логистики.
- **Золотодобыча** останется защитным активом, цены на золото прогнозируются на высоком уровне.
- **Удобрения** (URKZ, частично ALRS) выиграют от глобального спроса на продовольствие.
- **Уголь** (UGLD) – спрос со стороны Азии сохранится, но ограничения по инфраструктуре сдерживают рост.
- **Риски:** высокая ключевая ставка, волатильность рубля, санкции, изменение экспортных пошлин.
