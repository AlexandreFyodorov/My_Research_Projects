**Дата анализа:** 09 марта 2026 г.  
**Выборка:** 17 компаний, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100, позиции 31–60 по обороту).  
**Фавориты (выбраны пользователем):**  
- Группа 1 (отрицательный чистый долг, высокая дивидендная доходность): **LNZL, NLMK, MAGN**.  
- Группа 2 (компании с хорошими ROE/ROA, но с долговой нагрузкой, в порядке перечисления): **ALRS, SELG, CHMF, UGLD, MTLR, PLZL, GMKN, URKZ, BRZL**.

---


## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний, отобранных пользователем.  
Основные критерии:
- для первой группы: отрицательный чистый долг (NetDebt < 0) и наилучшая дивидендная доходность;
- для второй группы: высокие показатели рентабельности (ROE, ROA), но наличие положительного чистого долга, рассматриваемые в контексте долговой нагрузки.

В работе использованы данные из предоставленного файла. Большинство компаний относятся к металлургическому и горнодобывающему секторам, что позволяет проводить корректные сравнения внутри отрасли.

## 2. Общая характеристика сектора

Для сырьевых компаний ключевыми факторами инвестиционной привлекательности являются:
- **Долговая нагрузка (NetDebt/EBITDA)** – низкий долг повышает устойчивость к ценовым колебаниям;
- **Рентабельность (ROE, ROA)** – эффективность использования капитала и активов;
- **Дивидендная доходность** – важна для инвесторов, ориентированных на доход;
- **Мультипликаторы (P/E, P/B, EV/EBITDA)** – оценка рыночной привлекательности.

## 6. Выводы и рекомендации

### 6.1. Фавориты первой группы
- **NLMK** – наиболее сбалансированный выбор: низкий долг, высокая дивидендная доходность, крупная капитализация, низкие мультипликаторы. Рекомендуется для консервативных инвесторов.
- **LNZL** – экстремально высокая дивидендная доходность при отсутствии долга, но малая ликвидность и дорогие мультипликаторы создают риски. Подходит только для диверсифицированных портфелей с готовностью к высокой волатильности.
- **MAGN** – несмотря на отрицательный долг, убыточность и отрицательный P/E делают его нежелательным для долгосрочных вложений. Возможен лишь как спекулятивный актив на улучшение конъюнктуры.

### 6.2. Вторая группа (с долговой нагрузкой)
Наиболее привлекательные среди них:
- **URKZ** – исключительно низкие P/E и P/B, минимальный долг, высокая рентабельность. Отсутствие дивидендов компенсируется потенциалом роста котировок.
- **GMKN** – стабильный дивидендный плательщик с хорошей рентабельностью и умеренным долгом.
- **CHMF** – высокие дивиденды при низком долге, но дорогой P/E и низкая рентабельность капитала ограничивают потенциал.
- **ALRS** – умеренные показатели, дисконт к балансу, подходит для инвесторов, ищущих недооценённые активы.

Высокорискованные: **MTLR** (критический долг, отрицательный капитал), **SELG** (высокий долг и P/B), **PLZL** (экстремальный P/B). Эти компании требуют особой осторожности.

### 6.3. Общий вердикт
Пользователь сформировал качественную выборку. Наиболее перспективными с точки зрения соотношения риск/доходность выглядят **NLMK**, **URKZ**, **GMKN**. Остальные компании могут быть использованы для диверсификации или краткосрочных спекуляций при глубоком анализе их бизнес-моделей и перспектив.
