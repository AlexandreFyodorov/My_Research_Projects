# Аналитический отчёт по результатам фундаментального скрининга финансового сектора

**Дата анализа:** 17 марта 2026 г.  
**Выборка:** 6 компаний финансового сектора (банки и инвестиционный холдинг), прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **SFIN, VTBR, BSPB, SBER, MBNK, SVCB**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности ведущих компаний российского финансового сектора, представленных в файле и отобранных пользователем. В выборку вошли крупнейшие системообразующие банки, региональный банк с устойчивой бизнес-моделью, диверсифицированный инвестиционный холдинг, а также дочерний банк телекоммуникационной компании (МТС Банк).

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

**Анализ динамики:**  
Холдинг демонстрирует впечатляющий рост рентабельности с 2022 года. ROE увеличился с отрицательных значений в 2020 году до 23.9% в 2023-2024 годах, что свидетельствует о высокой эффективности управления портфелем активов. P/E снизился до привлекательных уровней (3.07-4.11), указывая на потенциальную недооценку. P/B также находится на минимальных уровнях (0.46-0.74), что говорит о дисконте к балансовой стоимости.

**Свежие данные и перспективы:**  
По данным LTM, дивидендная доходность обыкновенных акций составляет 115.4% – экстремально высокий показатель, связанный с разовыми выплатами или особенностями корпоративной структуры. P/E 4.11 и P/B 0.46 подтверждают глубокую недооценку. Рентабельность капитала (11.1%) остаётся на здоровом уровне.

**Вывод по тикеру:**  
SFIN – уникальный инвестиционный холдинг с экстремально высокой дивидендной доходностью и низкими мультипликаторами. Текущая оценка указывает на глубокую недооценку. Рекомендуется для дивидендных инвесторов с пониманием специфики холдинговой структуры.

**Анализ динамики:**  
ВТБ демонстрирует высокую волатильность финансовых показателей. В 2022 году банк получил убыток (отрицательный P/E), однако в 2023-2024 годах рентабельность капитала восстановилась до 22-23%. P/B находится на минимальных уровнях (0.35-0.4), что указывает на глубокую недооценку относительно капитала. P/E также держится на привлекательных уровнях (1.8-2.3).

**Свежие данные и перспективы:**  
По данным LTM, дивидендная доходность обыкновенных акций составляет 29.5%, что является очень высоким показателем. P/E 2.26 и P/B 0.4 подтверждают крайнюю недооценку. Рентабельность капитала (17.3%) остаётся высокой, несмотря на некоторое снижение с пиковых значений.

- Прогноз менеджмента по прибыли на 2025-2026 годы указывает на продолжение роста.
- Дивидендная политика ориентирована на выплату 25-50% чистой прибыли.
- Повышение требований к достаточности капитала с 2026 года может потребовать докапитализации.

**Вывод по тикеру:**  
ВТБ – высокорисковый, но потенциально высокодоходный актив. Низкая оценка и высокая дивидендная доходность создают потенциал роста. Рекомендуется для инвесторов с высокой толерантностью к риску.

Несмотря на временное снижение рентабельности в LTM, дивидендные перспективы поддерживают интерес к акциям. Рекомендуется для консервативных дивидендных инвесторов.

- Высокая достаточность капитала позволяет комфортно проходить регуляторные требования.

**Вывод по тикеру:**  
Сбер – лучший выбор для консервативных инвесторов в финансовом секторе. Устойчивые темпы роста, высокая рентабельность, развитие технологий и прогнозируемые дивиденды создают привлекательный инвестиционный профиль. Рекомендуется как якорный актив в портфеле.

## 5. Выводы и рекомендации

### 5.1. Лидеры по дивидендной доходности (фактической и прогнозной)

- **SFIN** – абсолютный лидер по фактической дивидендной доходности (115.4%). Экстремально высокая доходность требует проверки устойчивости выплат, но текущие мультипликаторы (P/E 4.1, P/B 0.46) подтверждают глубокую недооценку.


- **VTBR** – высокая фактическая дивидендная доходность (29.5%) при крайне низкой оценке (P/B 0.4). Высокорисковый, но потенциально высокодоходный актив.


- **SBER** – прогнозная дивидендная доходность 12.4% при устойчивых темпах роста и высокой рентабельности. Лучший выбор для консервативных инвесторов.


- **BSPB** – умеренная фактическая доходность (5%), но с потенциалом роста до двузначных значений при сохранении дивидендной политики.



### 5.2. Наиболее привлекательные по фундаментальным показателям

- **SBER** – лучшие показатели рентабельности капитала (19.9%) и активов (2.4%) среди всех банков выборки. Устойчивый рост и развитие технологий создают прочный фундамент.


- **MBNK** – низкая оценка (P/B 0.41) и высокая рентабельность (ROE 19%) при поддержке материнской структуры. Перспективный актив для роста капитала.


- **VTBR** – наиболее низкая оценка (P/B 0.4) создаёт потенциал для кратного роста при улучшении макроэкономической ситуации.



### 5.3. Умеренный интерес

- **SVCB** – привлекательная оценка (P/B 0.73), но более высокий P/E (6.1) и отсутствие дивидендов. Требует контроля за достаточностью капитала и восстановлением процентной маржи.



## 6. Перспективы сектора на 2026-2029 гг.

- **Ключевая ставка ЦБ** – ожидается дальнейшее снижение, что позитивно для банковского сектора: снизятся расходы на фондирование, вырастет чистая процентная маржа.
- **Регуляторные требования** – повышение минимальной достаточности капитала до 9.75% с 2026 года создаёт давление на банки с низким запасом прочности (VTBR, SVCB).
- **Технологическое развитие** – Сбер и МТС Банк активно развивают цифровые сервисы, что может стать драйвером роста комиссионных доходов.
- **Консолидация сектора** – крупнейшие банки продолжат укреплять позиции, региональные игроки (BSPB) сохранят нишевые преимущества.
- **Дивидендные выплаты** – финансовый сектор остаётся одним из самых щедрых по дивидендной доходности: прогнозы на 2025-2026 годы указывают на двузначные показатели у большинства эмитентов выборки.

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- динамику достаточности капитала VTBR и SVCB;
- реализацию дивидендных прогнозов по итогам 2025 года;
- финансовые результаты Сбера и подтверждение темпов роста прибыли;
- развитие синергии МТС Банка с головной группой;
- макроэкономические сигналы (ключевая ставка ЦБ, инфляция, объёмы кредитования).
