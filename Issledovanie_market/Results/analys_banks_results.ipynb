# Аналитический отчёт по результатам фундаментального скрининга российских акций
**Дата анализа:** 10 марта 2026 г.  
**Выборка:** 15 компаний, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100, позиции 31–60 по обороту).  
**Фавориты (выбраны пользователем):** SFIN, VTBR, BSPB, SBER, MBNK, SVCB.

---



## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний, отобранных пользователем в качестве фаворитов:  
**SFIN, VTBR, BSPB, SBER, MBNK, SVCB**.  

В ходе работы использованы данные из предоставленного файла. Следует учитывать, что часть показателей содержит некорректные значения (большие числа, похожие на коды), поэтому выводы делаются преимущественно на основе доступных рыночных мультипликаторов, дивидендной доходности, долговой нагрузки и оценок корпоративного управления. Все рассматриваемые компании относятся к **финансовому сектору** (банки и финансовые холдинги).

## 2. Общие характеристики сектора

Для финансовых организаций традиционные мультипликаторы **EV/EBITDA**, **NetDebt/EBITDA** не применимы (в файле они отсутствуют). Основное внимание уделялось:

- **P/E** – оценка окупаемости;
- **P/B** – соотношение цены и балансовой стоимости;
- **Дивидендная доходность** – привлекательность для инвесторов;
- **Чистый долг** – долговая нагрузка;
- **Рейтинги IR и прозрачности** (где представлены).

Все компании демонстрируют **отрицательный чистый долг** (денежные средства превышают долговые обязательства), что указывает на высокую ликвидность и минимальные риски дефолта.

## 5. Итоговые рекомендации

На основе имеющихся данных можно выделить следующие предпочтения:

1. **VTBR** – безусловный лидер по совокупности факторов: низкие мультипликаторы, высочайшая дивидендная доходность, отличная прозрачность. Интегральный рейтинг подтверждает это.
2. **SBER** – наиболее надёжный выбор для консервативных инвесторов: стабильные дивиденды, ликвидность, прозрачность, системная значимость.
3. **BSPB** – интересный вариант для тех, кто ищет сочетание низкой оценки и высокой дивидендной доходности, но с меньшей информационной открытостью.
4. **SFIN** – спекулятивный интерес из-за аномальной дивидендной доходности, но отсутствие IR-данных требует осторожности.
5. **MBNK** – неплохой дисконт и рентабельность, но низкая дивидендная доходность и непрозрачность снижают привлекательность.
6. **SVCB** – крепкий банк с хорошей рентабельностью, но оценка выше среднего и низкие дивиденды делают его аутсайдером среди фаворитов.

## 6. Ограничения анализа

- Часть показателей (ROE, ранги) в файле представлена некорректными значениями, что не позволило провести полноценное сравнение по рентабельности.
- Отсутствие данных по EBITDA и операционной прибыли у банков – нормально, но для небанковских структур (SFIN) это минус.
- Высокая дивидендная доходность VTBR и SFIN может быть следствием разовых факторов или ожиданий снижения прибыли – требуется проверка отчётности.

**Общий вывод:** Наиболее сбалансированными выглядят **VTBR** и **SBER**. Остальные компании могут быть интересны при углублённом анализе их бизнес-моделей и перспектив. Пользователь сформировал качественный список, который подтверждается данными файла.
