# Аналитический отчёт по результатам фундаментального скрининга сектора здравоохранения

**Дата анализа:** 17 марта 2026 г.  
**Выборка:** 9 компаний сектора здравоохранения (фармацевтика, медицинские услуги, биотехнологии), прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **GEMC, MDMG, DIOD, ABIO, OZPH, PRMD, GECO, LIFE, APTK**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского сектора здравоохранения, представленных в файле и отобранных пользователем. В выборку вошли операторы частных медицинских клиник, производители фармацевтической продукции, биотехнологические компании, а также аптечные сети. Сектор характеризуется защитными свойствами, устойчивым спросом на медицинские услуги и лекарства, а также высоким потенциалом роста в сегменте биотехнологий.

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

## 4. Детальный анализ фаворитов

### 4.1. GEMC (Европейский Медицинский Центр)

**Динамика ключевых мультипликаторов (2020-2025):**

| Показатель | 2020 | 2021 | 2022 | 2023 | 2024 | LTM |
|------------|------|------|------|------|------|-----|

| P/B | 9.85 | 17.1 | 2.57 | 12.5 | 4.86 | 4.86 |
| EV/EBITDA | 12 | 10.3 | 4.99 | 7.63 | 5.28 | 5.28 |

| ROA, % | 23.9 | 21.3 | 9.9 | 27.7 | 26.7 | 26.7 |

**Анализ динамики:**  
Компания демонстрирует высокую рентабельность капитала (ROE 37.1%) при комфортной долговой нагрузке. Чистый долг в LTM стал отрицательным, что указывает на наличие денежной подушки. P/E снизился до привлекательных 6.6, что может указывать на недооценку. P/B остаётся высоким (4.86), что характерно для компаний с сильными нематериальными активами.

**Свежие данные и перспективы:**  
Финансовые показатели компании стабильны, но выручка за последние 5 лет практически не растёт, оставаясь на уровне около 25 млрд руб., что при накопленной инфляции указывает на стагнацию. Капитальные затраты снижаются, что говорит об остановке развития. У компании отрицательный чистый долг, что создаёт возможность для инвестиций или дивидендов, однако дивиденды не выплачиваются с 2020 года. Менеджмент подтверждает намерение выплачивать дивиденды, но пока без конкретных действий.

**Вывод по тикеру:**  
GEMC – компания с сильным балансом и высокой рентабельностью, но отсутствие роста и дивидендов делает её малопривлекательной. Потенциальный возврат к дивидендным выплатам может стать драйвером роста. Рекомендуется для наблюдения.

Компания сталкивается с высокой конкуренцией, регуляторными ограничениями цен на лекарства и давлением онлайн-сервисов. Стратегия развития включает франчайзинг, развитие e-commerce и расширение ассортимента, но финансовые результаты пока слабые.

**Вывод по тикеру:**  
APTK – сложный актив с низкой рентабельностью и высокой оценкой. Бизнес сталкивается с серьёзными вызовами. Рекомендуется только для спекулятивных инвесторов с пониманием отраслевых рисков.

## 5. Выводы и рекомендации

### 5.1. Лидеры по дивидендной доходности

- **MDMG (Мать и дитя)** – дивидендная доходность 5.54% при высоких темпах роста (прогноз +28% в 2025, +23% в 2026) и устойчивом балансе. Лучший выбор для инвесторов, ориентированных на сочетание роста и дивидендов.


- **DIOD (ДИОД)** – высокая дивидендная доходность (7.24%), но убыточность ставит под сомнение устойчивость выплат. Высокорисковый актив.



### 5.2. Наиболее привлекательные по фундаментальным показателям

- **MDMG (Мать и дитя)** – сильные темпы роста, высокая рентабельность, отрицательный долг, щедрая дивидендная политика.


- **ABIO (Артген биотех)** – быстрый рост выручки и прибыли, минимальный долг, перспективы в биотехе.


- **PRMD (Промомед)** – потенциал кратного роста при успешном выводе новых препаратов (ВИЧ, онкология).


### 5.3. Умеренный интерес

- **GEMC (ЕМС)** – сильный баланс, но отсутствие роста и дивидендов. Потенциальный возврат к выплатам может стать драйвером.
- **OZPH (Озон Фармацевтика)** – стабильный бизнес с умеренным ростом, справедливая оценка.

### 5.4. Высокорисковые / требующие осторожности

- **DIOD (ДИОД)** – убытки, нестабильные показатели, риски сокращения дивидендов.
- **GECO (Генетико)** – убыточность, хотя динамика улучшается, высокие риски биотех-стартапов.
- **LIFE (LIFE)** – хроническая убыточность, неопределённые перспективы.
- **APTK (Аптечная сеть 36,6)** – низкая рентабельность, высокая оценка, отраслевые риски.

## 6. Перспективы сектора на 2026-2029 гг.

- **Частная медицина** (GEMC, MDMG) – будет расти за счёт оттока пациентов из государственной системы и развития платных услуг. Консолидация рынка через M&A (как в случае MDMG) создаёт дополнительный потенциал.
- **Фармацевтика** (OZPH, PRMD) – устойчивый спрос, импортозамещение, поддержка государства. Производители дженериков сохранят стабильность, инновационные компании (PRMD) могут показать взрывной рост при успешном выводе новых препаратов.
- **Биотехнологии** (ABIO, GECO) – наиболее динамичный сегмент с высоким потенциалом, но и высокими рисками. Развитие генной терапии, персонализированной медицины, импортозамещение создают долгосрочные драйверы.
- **Аптечный ритейл** (APTK) – сталкивается с давлением онлайн-каналов и регуляторными ограничениями. Консолидация и развитие омниканальности – ключевые факторы выживания.
- **Ключевые драйверы:** рост доходов населения, старение населения, импортозамещение в фармацевтике, господдержка биотеха, развитие ДМС.
- **Риски:** регуляторные изменения, контроль цен, высокая конкуренция, длительные циклы разработки и регистрации новых препаратов.

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- финансовые результаты MDMG после интеграции приобретённых активов и подтверждение прогнозов роста;
- появление дивидендных выплат у GEMC;
- прогресс в разработке и регистрации новых препаратов PRMD (ВИЧ, онкология);
- динамику убыточности и выход на прибыль GECO;
- способность ABIO сохранять высокие темпы роста;
- макроэкономические сигналы (доходы населения, инфляция, регуляторные изменения в фармацевтике).
