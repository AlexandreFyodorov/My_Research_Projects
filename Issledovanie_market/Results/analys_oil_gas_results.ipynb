# Аналитический отчёт по результатам фундаментального скрининга российских акций
**Дата анализа:** 09 марта 2026 г.  
**Выборка:** 17 компаний, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100, позиции 31–60 по обороту).  
**Фавориты (выбраны пользователем):** SNGSP, BANEP, LKOH, ROSN, NVTK, TATNP.

---



---

## 1. Введение
В результате выполнения автоматизированного скрипта фундаментального анализа для каждой компании рассчитаны ключевые финансовые мультипликаторы, показатели эффективности, долговой нагрузки, дивидендной политики, а также построен интегральный рейтинг (`Composite_Rank`). Чем меньше значение рейтинга, тем более привлекательной считается компания по совокупности факторов (низкая оценка, высокая рентабельность, низкий долг, высокая дивидендная доходность).

Данный отчёт содержит подробный разбор шести компаний, отобранных пользователем как фавориты, и обоснование их инвестиционной привлекательности.

## 4. Детальный разбор фаворитов

### 4.1. SNGSP (Сургутнефтегаз, привилегированные акции)
**Ключевые показатели:**
- **Капитализация:** 1152 млрд руб.
- **P/E:** 0.87 (минимальное значение в выборке)
- **P/B:** 0.16 (огромный дисконт к балансовой стоимости)
- **ROE:** 18.7% (высокая рентабельность)
- **Дивидендная доходность:** 16.19% (максимальная среди фаворитов)
- **NetDebt/EBITDA:** –9.39 (огромная чистая денежная позиция: 5967 млрд руб. денег против нулевого долга)
- **FCF:** 622.5 млрд руб.
- **Composite_Rank:** 23.5 (лучший в выборке)

**Сильные стороны:**
- Уникальная финансовая структура – компания-«денежный мешок».
- Экстремально низкие мультипликаторы указывают на глубокую недооценку.
- Высокая дивидендная доходность при стабильной прибыли.
- Лучший интегральный рейтинг.

**Риски:**
- Зависимость от цен на нефть.
- Непрозрачность использования гигантских денежных запасов.

## 5. Выводы и инвестиционные соображения

1. **SNGSP** – уникальный случай «денежного мешка» с экстремально низкой оценкой и высокой дивидендной доходностью. Безусловный лидер по интегральному рейтингу. Рекомендуется для инвесторов, ориентированных на глубокую стоимость и дивиденды.

2. **BANEP** и **LKOH** – отличные дивидендные истории с низким долгом и привлекательными мультипликаторами. LKOH – более крупный и диверсифицированный, BANEP – более высокая доходность при меньшем размере.

3. **TATNP** – сочетает высокую рентабельность капитала, умеренную оценку и хорошую дивидендную доходность. Сбалансированный выбор.

4. **ROSN** – крупнейшая компания с нулевым долгом и низкой оценкой, но уступает по рентабельности и дивидендам. Может быть интересна консервативным инвесторам.

5. **NVTK** – история роста с премией в цене. Подойдёт тем, кто делает ставку на развитие СПГ и готов мириться с более высокой оценкой.

Все выбранные фавориты имеют **низкий или отрицательный чистый долг**, что минимизирует кредитные риски. Представленный набор охватывает разные инвестиционные стили и позволяет сформировать диверсифицированный портфель с акцентом на нефтегазовый сектор – основу российского рынка.

*Рекомендуется дополнительно учитывать макроэкономические факторы (цены на нефть, курсы валют) и дивидендную политику каждой компании.*
