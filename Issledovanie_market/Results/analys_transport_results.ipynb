# Аналитический отчёт по результатам фундаментального скрининга (Транспорт и Логистика)

**Дата анализа:** 18 марта 2026 г.
**Выборка:** 10 компаний из сектора транспорта и логистики, представленных в файле.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского транспортно-логистического сектора. Выборка включает морские перевозки (FLOT, NMTP), авиаперевозки (AFLT), автомобильные грузоперевозки (GTRK, DELI), производство транспортных средств (KMAZ), речные перевозки (FESH, UWGN), а также операторов микромобильности (WUSH) и производителя спецодежды (SVAV), чья деятельность тесно связана с промышленным сектором. Сектор характеризуется высокой зависимостью от макроэкономической конъюнктуры, цен на топливо, инвестиционной активности и геополитических факторов.

В работе использованы исторические данные из предоставленного файла (2020–2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

**Анализ динамики:**
Компания показала сильные результаты в 2021–2023 гг., воспользовавшись высокими ставками фрахта. Рентабельность капитала достигала 19%. Долговая нагрузка остается низкой, а P/B на уровне 0.6 указывает на потенциальную недооценку, но с учетом убыточности.

**Свежие данные и перспективы:**

Это расширит возможности компании в перспективном сегменте транспортировки СПГ. Рынок специализированных морских перевозок остается волатильным, но компания сохраняет лидирующие позиции.

**Вывод по тикеру:**
Текущая убыточность делает оценку сложной. Однако сильный баланс (низкий долг) и обновление флота создают потенциал для восстановления прибыльности при улучшении рыночной конъюнктуры. Подходит для инвесторов с высоким риском, ожидающих разворота цикла в морских перевозках.

## 5. Выводы и рекомендации по сектору

### 5.1. Наиболее привлекательные по фундаментальным показателям

- **NMTP (НМТП)** – абсолютный лидер в выборке: высокая ROE (21%), отрицательный чистый долг (нетто-кэш), низкие мультипликаторы (P/E 4.34) и щедрая дивидендная доходность (11.1%). Рекомендуется для всех типов инвесторов.


- **AFLT (Аэрофлот)** – успешная история восстановления. Рост операционных показателей, снижение долга, выход на высокую рентабельность и дивиденды. Рекомендуется к покупке.


### 5.2. Умеренный интерес / Спекулятивные идеи

- **FLOT (Совкомфлот)** – временно убыточен, но имеет низкий долг и обновляет флот. Интересен для инвесторов, готовых ждать восстановления цикла.


- **FESH (FESCO)** – стабильный игрок с низким долгом, но с нормализовавшейся рентабельностью. Подходит для долгосрочного удержания.


- **WUSH (Whoosh)** – находится в фазе перехода от роста к эффективности. Рост долга и убытки настораживают. Стоит дождаться плана по повышению эффективности.


- **SVAV (СОЛЛЕРС)** – финансово устойчив, но снижение ROE и рост P/E снижают привлекательность.


### 5.3. Высокорисковые / Требующие осторожности

- **DELI (Делимобиль)** – высокий долг, убыточность, проблемы с рефинансированием. Не рекомендуется.
- **KMAZ (КАМАЗ)** – критическое финансовое состояние из-за высоких инвестрасходов. Поддержка государства — единственный якорь надежды. Только для самых рискованных.
- **UWGN (UWC)** – цикличный убыточный бизнес с высоким долгом. Не рекомендуется.
- **GTRK (Globaltruck)** – катастрофическое состояние, риск банкротства ключевой "дочки". Категорически не рекомендуется.

## 6. Перспективы сектора на 2026–2029 гг.

- **Морские перевозки и порты (FLOT, NMTP):** Рынок ставок фрахта может оставаться волатильным. Ключевыми драйверами станут переориентация грузопотоков на Восток и Юг, развитие инфраструктуры и обновление флота. НМТП выглядит более устойчиво за счет диверсификации грузов.
- **Авиаперевозки (AFLT):** Восстановление международного сообщения и внутренний туризм будут поддерживать спрос. Ключевой риск — обслуживание и лизинг оставшегося парка иностранных самолетов.
- **Автомобильные и ж/д перевозки (FESH, GTRK, DELI):** Зависят от деловой активности. Высокая ключевая ставка будет давить на инвестиции в автопарки и, соответственно, на операторов. Контейнеризация продолжит поддерживать FESCO.
- **Машиностроение (KMAZ, UWGN):** Циклические отрасли, находящиеся под давлением из-за высокой ставки и насыщения рынка. Восстановление возможно только после начала цикла снижения ставок и обновления парков.
- **Микромобильность (WUSH):** Рынок продолжит расти, но конкуренция и регуляторика будут усиливаться. Ключевым станет умение компаний балансировать между ростом и рентабельностью.
- **Общие риски:** Высокая ключевая ставка ЦБ, инфляция издержек (топливо, запчасти, лизинг), геополитическая напряженность, влияющая на логистические маршруты, риски банкротства слабых игроков.

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- исполнение дивидендных ожиданий NMTP и AFLT;
- публикацию полугодовых отчетов для подтверждения трендов у лидеров;
- выход на операционную прибыльность FLOT и WUSH;
- разрешение ситуации с долгами DELI и банкротством GTRK;
- макроэкономические сигналы (ставка ЦБ, инфляция, динамика грузооборота);
- прогресс в реализации стратегических инвестпроектов KMAZ.
