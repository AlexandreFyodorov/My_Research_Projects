# Аналитический отчёт по результатам фундаментального скрининга (девелопмент, IT, финансы)

**Дата анализа:** 18 марта 2026 г.  
**Выборка:** 9 компаний из секторов девелопмента, IT и финансов, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **CNRU, LSRG, SMLT, CARM, PIKK, MSTT, DOMRF, BAZA, GLRX**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний из различных секторов экономики, представленных в файле и отобранных пользователем. В выборку вошли представители IT-сектора (Циан, Группа «Базис»), крупнейшие девелоперы (Группа ЛСР, ПИК, Самолёт, GloraX), финансовый институт развития (ДОМ.РФ), а также микрофинансовая организация (CarMoney). В связи с неоднородностью выборки анализ проводился с учётом специфики каждого бизнеса.

В работе использованы данные из предоставленного файла (исторические данные за 2020-2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

### 4.1. CNRU (Циан)

**Анализ динамики:**  
Компания демонстрирует уверенный рост после редомициляции. P/E снизился с пиковых значений до 16.8. Рентабельность капитала стабильно высокая (41.4%). Долговая нагрузка отсутствует.

**Свежие данные и перспективы:**  
По итогам 3 квартала 2025 года совокупная выручка выросла на 29% до 4,1 млрд руб., скорректированная EBITDA увеличилась более чем в 2 раза, рентабельность по EBITDA достигла 28%. Совет директоров рекомендовал специальный дивиденд 104 руб. на акцию (доходность ~17-18%). Следующий дивиденд планируется в середине 2026 года в размере более 50 руб. на акцию. Компания полностью свободна от долга и обладает денежной подушкой более 10 млрд руб.

**Вывод по тикеру:**  
CNRU – один из самых привлекательных активов в выборке, сочетающий технологический рост, отсутствие долга и регулярные дивиденды. Рекомендуется как для дивидендной стратегии, так и для долгосрочного портфеля.

## 5. Выводы и рекомендации

### 5.1. Наиболее привлекательные по фундаментальным показателям

- **CNRU (Циан)** – лучший выбор в выборке: отсутствие долга, высокая рентабельность (ROE 41.4%), дивидендная доходность 14.1% и программа buyback. Рекомендуется для всех типов инвесторов.


- **DOMRF (ДОМ.РФ)** – наиболее надежный актив с государственным участием (P/E 4.45, P/B 0.84). Рекомендуется для консервативных инвесторов.


- **LSRG (Группа ЛСР)** – устойчивый дивидендный актив (доходность 8.97%) с недооценкой по P/B (0.62) и умеренным долгом.


- **BAZA (Группа «Базис»)** – быстрорастущая IT-компания с высокими темпами роста (выручка +57%) и отрицательным долгом. Рекомендуется для инвесторов, ориентированных на рост капитала.



### 5.2. Умеренный интерес

- **PIKK (Группа ПИК)** – лидер рынка с отрицательным долгом, но под давлением макроэкономической среды. Ожидание снижения ключевой ставки как катализатор роста.


- **GLRX (GloraX)** – среднесрочная идея с гарантированной офертой 19% годовых и хорошими темпами роста.


### 5.3. Высокорисковые / требующие осторожности

- **SMLT (ГК Самолёт)** – критическая долговая нагрузка (NetDebt/EBITDA 3.10), процентные расходы съедают операционную прибыль.
- **CARM (СмартТехГрупп)** – непрозрачная структура, слабая операционная динамика, сомнения в достижении стратегических целей.
- **MSTT** – требует уточнения эмитента, возможное несоответствие тикеру.

## 6. Перспективы секторов на 2026-2029 гг.

- **IT-сектор** (CNRU, BAZA) – продолжит рост благодаря импортозамещению и цифровизации. Циан выигрывает от оживления рынка недвижимости.
- **Девелопмент** (LSRG, PIKK, GLRX, SMLT) – находится в фазе адаптации к высокой ключевой ставке. Снижение ставки ЦБ станет ключевым катализатором роста. Консолидация сектора продолжится.
- **Финансовый сектор с госучастием** (DOMRF) – сохранит стабильность и доступ к дешевому фондированию.
- **Риски:** сохранение жесткой ДКП дольше ожиданий, снижение реальных доходов населения, изменения в программах господдержки ипотеки, санкционное давление, корпоративные риски (SMLT, CARM).

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- реализацию дивидендных планов CNRU (ожидаемый дивиденд в середине 2026 года);
- динамику ключевой ставки и ее влияние на девелоперов (LSRG, PIKK, SMLT, GLRX);
- исполнение оферты GLRX для акционеров;
- финансовые результаты BAZA и подтверждение прогнозов роста;
- способность SMLT обслуживать долговую нагрузку в условиях сохранения высоких ставок;
- уточнение эмитента по тикеру MSTT;
- макроэкономические сигналы (ставка ЦБ, инфляция, объемы выдачи ипотеки).
