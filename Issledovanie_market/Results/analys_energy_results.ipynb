# Аналитический отчёт по результатам фундаментального скрининга (Электроэнергетика)

**Дата анализа:** 18 марта 2026 г.  
**Выборка:** 16 компаний российского сектора электроэнергетики, представленных в файле.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского сектора электроэнергетики. Выборка включает как сетевые компании (распределительные и магистральные), так и генерирующие активы (оптовые и территориальные генерирующие компании). Сектор характеризуется высокой степенью государственного регулирования тарифов, естественно-монопольным положением большинства компаний и, как следствие, стабильными, но не всегда высокими темпами роста. Ключевыми факторами для инвесторов здесь являются дивидендная доходность, долговая нагрузка и эффективность инвестиционных программ.

В работе использованы исторические данные из предоставленного файла (2020–2025 гг. и последние 12 месяцев – LTM), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года.

**Вывод по тикеру:**

## 5. Выводы и рекомендации по сектору

### 5.1. Наиболее привлекательные по фундаментальным показателям

- **MRKP (Россети Центр и Приволжья)** – лучший выбор в секторе: высокая и растущая ROE (16.5%), минимальный долг (0.15), P/B 0.6, высокая дивидендная доходность (8.7%). Рекомендуется для всех типов инвесторов.


- **MSRS (Россети Московский Регион)** – качественный сетевой актив со стабильным ростом, низким долгом и недооценкой (P/B 0.34). Дивидендная доходность 9.1%. Рекомендуется для долгосрочных инвестиций.


- **IRAO (Интер РАО)** – финансовая устойчивость (нетто-кэш), стабильная прибыльность (ROE 11.7%) и крайне низкие мультипликаторы (P/E 2.53). Риск — дивидендная политика.


- **MRKV (Россети Волга)** – отличная динамика 2025 года, низкий долг, дивидендная доходность 7.8%. Рекомендуется к покупке.


### 5.2. Умеренный интерес / Дивидендные истории

- **MSNG (Мосэнерго)** – исключительно для дивидендных инвесторов (доходность 10.2%) при низком качестве роста (ROE 0.5%).


- **MRKC, MRKU** – хорошие дивидендные истории с ROE 12–13% и низким долгом, но уступают лидерам (MRKP, MSRS) по темпам роста и рентабельности.


- **TGKA (ТГК-1)** – глубокая недооценка (P/B 0.19) и стабильные дивиденды, но бизнес сталкивается с регуляторными и рыночными вызовами.


- **UPRO (Юнипро)** – фундаментально дешева, но заблокирована для инвесторов из-за геополитических рисков. Только для рискованных спекуляций на снятии блокировки.


### 5.3. Высокорисковые / Требующие осторожности

- **FEES (Россети)** – убыточность и неясные перспективы восстановления. Не рекомендуется.
- **TGKN (ТГК-14)** – несмотря на рост ROE, ухудшение текущих результатов и корпоративный скандал повышают риски.
- **MRKZ (Россети Юг)** – слабая рентабельность и отсутствие дивидендов.
- **OGKB (ОГК-2)** – дефицит информации для анализа.

## 6. Перспективы сектора на 2026–2029 гг.

- **Сетевые компании (FEES, MRKP, MSRS, MRKV, MRKC, MRKU, LSNG, MRKZ):** Будут оставаться под влиянием тарифного регулирования. Ключевым драйвером роста станет индексация тарифов на передачу, которая должна превышать инфляцию для финансирования инвестпрограмм. Снижение ключевой ставки снизит стоимость обслуживания долга. Наибольший потенциал у компаний с низким долгом и высокой эффективностью (MRKP, MSRS).
- **Генерирующие компании (IRAO, MSNG, UPRO, HYDR, TGKA, TGKN, OGKB):** Рынок электроэнергии и мощности будет зависеть от роста энергопотребления со стороны бизнеса и населения. Программы ДПМ (договоры предоставления мощности) для многих ТЭС завершаются, что может снизить денежный поток, но высвободить средства для дивидендов. Интер РАО выигрывает от трейдинга и экспорта. Риски сохраняются в виде роста цен на топливо (газ, уголь) и ужесточения экологического регулирования.
- **Общие риски:** Регуляторные изменения, уровень инфляции и ключевой ставки ЦБ, геополитическая ситуация, влияющая на экспорт и санкции в отношении отдельных компаний, рост неплатежей потребителей.

## 7. Когда произвести очередной скрининг?

Рекомендуется провести следующий скрининг **через 6 месяцев (сентябрь 2026)** с акцентом на:

- утверждение и реализацию дивидендных рекомендаций компаний, особенно IRAO и HYDR;
- публикацию полугодовых отчетов по МСФО за 2026 год для подтверждения трендов роста у MRKP, MSRS, MRKV;
- макроэкономические сигналы (ставка ЦБ, инфляция, динамика энергопотребления);
- корпоративные события в TGKN и OGKB;
- разрешение геополитической ситуации вокруг UPRO.
