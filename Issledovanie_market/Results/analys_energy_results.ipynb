# Аналитический отчёт по результатам фундаментального скрининга сектора Электроэнергетики

**Дата анализа:** 14 марта 2026 г.  
**Выборка:** 23 компании сектора электроэнергетики, прошедших фильтр ликвидности (оборот ≥ 1 млн руб., заявки ≥ 100).  
**Фавориты (выбраны пользователем):** **FEES, OGKB, TGKA, MRKV, IRAO, MSNG, TGKN, MSRS, MRKP, MRKC, MRKU, LSNG, UPRO, HYDR, MRKZ**.

## 1. Введение

Цель анализа – оценка инвестиционной привлекательности компаний российского электроэнергетического сектора, представленных в файле и отобранных пользователем для всестороннего рассмотрения. В выборку вошли крупнейшие представители отрасли: сетевые компании (распределительные и магистральные), генерирующие компании (оптовые и территориальные), а также энергосбытовые структуры.

В работе использованы данные из предоставленного файла (скрининг от 14 марта 2026 г.), а также актуальная информация из открытых источников для оценки перспектив компаний на ближайшие 3 года. Сектор электроэнергетики характеризуется высокой регуляторной защитой, стабильными денежными потоками и дивидендными выплатами, что делает его привлекательным для консервативных инвесторов.

**Анализ ситуации и перспективы:**  
Россети (ФСК) – системообразующая сетевая компания, контролирующая магистральные линии электропередачи. В рамках реформы электросетевого комплекса продолжается консолидация распределительных сетей под управлением холдинга [citation:3]. **Ключевые факторы на 2026-2029 гг.:**
- Государственное регулирование тарифов обеспечивает стабильность, но ограничивает рост.
- Программа модернизации сетей требует значительных капзатрат.
- Потенциальное ужесточение дивидендной политики при снижении инвестиционной нагрузки.

**Вердикт:** Консервативный актив для долгосрочных инвесторов, ориентированных на стабильность.

## 5. Выводы и рекомендации

### 5.1. Фавориты среди фаворитов

- **IRAO (Интер РАО)** – абсолютный лидер сектора: отсутствие долга, высокая рентабельность, дивидендная доходность 8-10% и потенциал роста котировок от текущих уровней (3.11 руб.) [citation:3].
- **UPRO (Юнипро)** – лидер по дивидендной доходности (10-12%) с низким долгом и высокой рентабельностью.
- **OGKB (ОГК-2)** – привлекательное сочетание дивидендов (8-10%), низкой оценки (P/B < 0.7) и умеренного долга.
- **MRKV, MRKP, MRKC (сетевые компании)** – стабильный дивидендный поток (8-10%) при низкой долговой нагрузке. Особый интерес представляет MRKC с ожиданиями роста дивидендов до 18-20 коп. [citation:5].

### 5.2. Дивидендные истории

- **MSRS, LSNG, MSNG** – стабильные выплаты (7-9%) при умеренной долговой нагрузке.
- **HYDR** – долгосрочный актив с потенциалом роста, но дивиденды ниже среднего по сектору.

### 5.3. Высокорискованные / требующие осторожности

- **TGKN** – малая капитализация, региональная зависимость.
- **FEES** – неопределённость дивидендной политики, высокие капзатраты.

## 6. Перспективы сектора на 2026-2029 гг.

- **Снижение ключевой ставки** (ожидания рынка – до 14-15% к концу 2026) [citation:5] станет ключевым катализатором роста для всего сектора, особенно для компаний с долговой нагрузкой.
- **Завершение инвестиционных программ ДПМ** высвободит средства для дивидендов у генерирующих компаний (ОГК-2, ТГК-1, Мосэнерго).
- **Консолидация сетевого комплекса** под управлением Россетей может привести к реорганизациям и выкупам акций (тема активно обсуждается для MRKC) [citation:5].
- **Цифровизация сетей** и повышение эффективности остаются приоритетами для сетевых компаний.
- **Риски:** государственное регулирование тарифов, рост стоимости топлива для тепловой генерации, волатильность водности для гидрогенерации.

- макроэкономические сигналы (инфляция, промпроизводство, энергопотребление).
