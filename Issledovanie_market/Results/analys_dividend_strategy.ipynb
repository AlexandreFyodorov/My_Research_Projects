# Финальный инвестиционный отчёт: Дивидендная стратегия

**Дата анализа:** 18 марта 2026 г.  
**Портфель:** Дивидендные акции российских компаний, отобранные по секторам.

## 1. Введение

Цель данного отчёта – сформировать рекомендации по инвестированию в дивидендные акции российских компаний на основе предоставленных фундаментальных данных (МСФО за 2020–2025 гг.) и текущей рыночной ситуации (данные QUIK, RSI). Пользователем был произведён предварительный отбор тикеров по секторам с учётом дивидендной доходности и технического индикатора RSI. В рамках анализа мы углублённо оценим каждую компанию с точки зрения долгосрочной дивидендной стратегии, проверим мультипликаторы (P/E, P/B, EV/EBITDA, ROE), долговую нагрузку и степень недооценённости/переоценённости на основе текущих цен и исторической динамики.

В работе использованы исторические данные из предоставленного файла (2020–2025 гг. и последние 12 месяцев – LTM), а также актуальные цены и уровень RSI (приблизительно) из файлов котировок. На основе совокупности факторов будут даны рекомендации по включению каждого тикера в дивидендный портфель.

```python
# %% [markdown]
# # Скрипт для отображения графиков цен и индикаторов
# 
# Данный скрипт загружает исторические данные по акциям из файлов формата CSV (экспорт из QUIK),
# строит графики цен, скользящие средние и индикатор RSI для выбранных тикеров.

# %% [markdown]
# ## 1. Установка необходимых библиотек (если ещё не установлены)

# %%
!pip install mplfinance  # раскомментируйте, если нужно установить

# %% [markdown]
# ## 2. Импорт библиотек

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
import os
import warnings
warnings.filterwarnings('ignore')

import mplfinance as mpf

# %% [markdown]
# ## 3. Настройка пути к данным и словарь соответствия

# %%
# Путь к папке с файлами (замените на ваш)
DATA_PATH = 'C:/Users/1/Documents/TRADE_2026/ScriptsAnalizTorgy/Quik/'

# Словарь соответствия: тикер -> имя файла
ticker_to_file = {
    'SNGSP': 'Сургнфгз-п [Price].txt',
    'BANEP': 'Башнефт ап [Price].txt',
    'LKOH': 'ЛУКОЙЛ [Price].txt',
    'TATNP': 'Татнфт 3ап [Price].txt',
    'ROSN': 'Роснефть [Price].txt',
    'NVTK': 'Новатэк ао [Price].txt',
    'VTBR': 'ВТБ ао [Price].txt',
    'BSPB': 'БСП ао [Price].txt',
    'SBER': 'Сбербанк [Price].txt',
    'SFIN': 'ЭсЭфАй ао [Price].txt',
    'PLZL': 'Полюс [Price].txt',
    'MTSS': 'МТС-ао [Price].txt',
    'TTLK': 'Таттел. ао [Price].txt',
    'T': 'Т-Техно ао [Price].txt',
    'MDMG': 'MDMG-ао [Price].txt',
    'NKHP': 'НКХП ао [Price].txt',
    'PHOR': 'ФосАгро ао [Price].txt',
    'AKRN': 'Акрон [Price].txt',
    'HEAD': 'Хэдхантер [Price].txt',
    'MGKL': 'МГКЛ [Price].txt',
    'OZON': 'Озон [Price].txt',
    'CNRU': 'Циан [Price].txt',
    'LSRG': 'ЛСР ао [Price].txt',
    'MSNG': 'МосЭнерго [Price].txt',
    'IRAO': 'ИнтерРАОао [Price].txt',
    'TGKN': 'ТГК-14 [Price].txt',
    'MSRS': 'РСетиМР ао [Price].txt',
    'MRKP': 'РСетиЦП ао [Price].txt',
    'MRKC': 'РоссЦентр [Price].txt',
    'MRKV': 'РсетВол ао [Price].txt',
    'MRKU': 'Россети Ур [Price].txt',
    'NMTP': 'НМТП ао [Price].txt',
    'AFLT': 'Аэрофлот [Price].txt',
    'SVAV': 'СОЛЛЕРС [Price].txt',
}

# %% [markdown]
# ## 4. Функции загрузки данных и расчёта индикаторов

# %%
def load_ticker_data(ticker):
    """Загружает данные для заданного тикера из файла."""
    filename = ticker_to_file.get(ticker)
    if filename is None:
        print(f"Предупреждение: для тикера {ticker} не задан файл.")
        return None
    
    filepath = os.path.join(DATA_PATH, filename)
    if not os.path.exists(filepath):
        print(f"Файл {filepath} не найден.")
        return None
    
    # Читаем CSV (без дополнительных параметров даты)
    df = pd.read_csv(filepath)
    # Переименовываем колонки: убираем угловые скобки и приводим к нижнему регистру
    df.columns = [col.strip('<').strip('>').lower() for col in df.columns]
    # Преобразуем столбец DATE в datetime
    df['date'] = pd.to_datetime(df['date'], format='%Y%m%d')
    df.set_index('date', inplace=True)
    # Оставляем только нужные колонки
    df = df[['open', 'high', 'low', 'close', 'vol']]
    # Сортируем по дате
    df.sort_index(inplace=True)
    return df

def add_sma(df, window=20):
    df[f'sma_{window}'] = df['close'].rolling(window=window).mean()
    return df

def add_rsi(df, window=14):
    delta = df['close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=window).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=window).mean()
    rs = gain / loss
    df['rsi'] = 100 - (100 / (1 + rs))
    return df

# %% [markdown]
# ## 5. Построение графиков для выбранных тикеров

# %%
tickers_to_plot = [
    'SNGSP', 'LKOH', 'VTBR', 'SBER', 'NMTP', 'CNRU', 'IRAO',  # ядро
    'BANEP', 'TATNP', 'BSPB', 'SFIN', 'MTSS', 'HEAD', 'LSRG', 'MSNG',  # вторичные
]

plt.style.use('seaborn-v0_8-darkgrid')
figsize = (14, 10)

for ticker in tickers_to_plot:
    print(f"Загрузка данных для {ticker}...")
    df = load_ticker_data(ticker)
    if df is None or df.empty:
        continue
    
    df = add_sma(df, window=20)
    df = add_sma(df, window=50)
    df = add_rsi(df, window=14)
    
    fig, axes = plt.subplots(3, 1, figsize=figsize, sharex=True,
                             gridspec_kw={'height_ratios': [3, 1, 1]})
    
    ax1 = axes[0]
    ax1.plot(df.index, df['close'], label='Close', color='black', linewidth=1)
    ax1.plot(df.index, df['sma_20'], label='SMA 20', color='blue', linestyle='--', linewidth=1)
    ax1.plot(df.index, df['sma_50'], label='SMA 50', color='red', linestyle='--', linewidth=1)
    ax1.set_ylabel('Цена')
    ax1.legend(loc='upper left')
    ax1.set_title(f'{ticker} — цена и индикаторы')
    ax1.grid(True)
    
    ax2 = axes[1]
    ax2.bar(df.index, df['vol'], color='gray', alpha=0.5, width=20)
    ax2.set_ylabel('Объём')
    ax2.grid(True)
    
    ax3 = axes[2]
    ax3.plot(df.index, df['rsi'], label='RSI 14', color='purple', linewidth=1)
    ax3.axhline(70, color='red', linestyle='--', linewidth=0.8, label='Overbought (70)')
    ax3.axhline(30, color='green', linestyle='--', linewidth=0.8, label='Oversold (30)')
    ax3.set_ylabel('RSI')
    ax3.set_ylim(0, 100)
    ax3.legend(loc='upper left')
    ax3.grid(True)
    
    for ax in axes:
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
        ax.xaxis.set_major_locator(mdates.YearLocator())
        ax.xaxis.set_minor_locator(mdates.MonthLocator())
    
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# %% [markdown]
# ## 6. Свечной график с mplfinance (опционально)

# %%
# Пример для одного тикера (убедитесь, что mplfinance установлен)
ticker_candle = 'SBER'
df_candle = load_ticker_data(ticker_candle)
if df_candle is not None:
    # Переименуем колонки для mplfinance (первая буква заглавная)
    df_candle.columns = ['Open', 'High', 'Low', 'Close', 'Volume']
    ap = [
        mpf.make_addplot(df_candle['Close'].rolling(20).mean(), color='blue', width=0.7),
        mpf.make_addplot(df_candle['Close'].rolling(50).mean(), color='red', width=0.7),
    ]
    mpf.plot(df_candle, type='candle', style='charles', volume=True, addplot=ap,
             title=f'{ticker_candle} — свечной график', figsize=figsize,
             datetime_format='%Y-%m', xrotation=45)
```
